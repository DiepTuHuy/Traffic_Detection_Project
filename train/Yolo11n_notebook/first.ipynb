```python
!pip install ultralytics
from ultralytics import YOLO
import torch
import os

def main():
    BATCH_SIZE = 16 

    WORKERS = 2 

    model = YOLO('yolo11n.pt') 

    results = model.train(
        data='/kaggle/input/datasets/hunghoang23123/tsd-yolo11n/data.yaml',
        epochs=100,
        imgsz=640,
        device=[0, 1],   # Chạy 2 GPU
        batch=BATCH_SIZE,
        workers=WORKERS, # Quan trọng: Giảm worker xuống
        project='traffic_sign_project',
        name='yolo11',
        exist_ok=True,
        amp=True,
        lrf=0.001
    )

if __name__ == '__main__':
    # Bắt buộc cho DDP
    torch.multiprocessing.set_start_method('spawn', force=True)
    main()
```
